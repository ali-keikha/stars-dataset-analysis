# Star Dataset Analysis

## Project Overview

This notebook explores the characteristics of M-Class stars from the Star Dataset. The analysis includes data cleaning, color distribution, luminosity, temperature, radius, and star type statistics to better understand the properties of this spectral class.

```python
import numpy as np
import pandas as pd
```

## Load the Dataset

```python
df = pd.read_csv("../data/Stars.csv.xls")
```

## Research Objective

### Analysis of M-Class Stars

The color labels contain inconsistent naming conventions. Before performing the analysis, these values are standardized to improve data quality and ensure accurate results.

## Data Cleaning

```python
df["Color"] = df["Color"].replace(
    {
    "Blue-white": "Blue White",
    "Blue white": "Blue White",
    "Blue-White": "Blue White",

    "white": "White",

    "yellowish": "Yellowish",

    "yellow-white": "Yellow White",

    "White-Yellow": "White Yellow",

    "Pale yellow orange": "Pale Yellow Orange",

    "Orange-Red": "Orange Red"
}
)
```

### Distribution of Star Colors

```python
class_m = df.query("Spectral_Class == 'M' ")

color_count = class_m["Color"].value_counts()
for index in color_count.index:
    print(f"{index} Count is: {color_count[index]}")
    print("-" * 25)
```

### Most and Least Common Colors

```python
print(f"The Most Common Color is: {color_count.idxmax()} with {color_count.max()} count")
print(f"The least common color is: {color_count.idxmin()} with {color_count.min()} count")
```

### Average Luminosity

```python
print(f"Average Luminosity of M-Class Stars: {np.round(class_m["L"].mean(), 3)} L")
```

### Average Temperature

```python
print(f"Average Temperature of M-Class Stars: {np.round(class_m["Temperature"].mean(), 3)} K")
```

### Luminosity and Temperature Range

```python
print("Luminosity Statistics")
print("-" * 25)
print(" " * 4 ,f"Maximum Luminosity: {class_m["L"].max()} L")
print(" " * 4 ,f"Minimum Luminosity: {class_m["L"].min()} L")
print()
print("Temperature Statistics")
print("-" * 25)
print(" " * 4, f"Maximum Temperature:: {class_m["Temperature"].max()} K")
print(" " * 4, f"Minimum Temperature: {class_m["Temperature"].min()} K")
```

### Average Radius

```python
print(f"Average Radius: {np.round(class_m["R"].mean(), 3)}")
```

### Radius Range

```python
print(f"Maximum Radius: {class_m["R"].max()}")
print(f"Minimum Radius: {class_m["R"].min()}")
```

### Distribution of Star Types

```python
type_count = class_m["Type"].value_counts()
for index in type_count.index:
    print(f"Type {index} Count: {type_count[index]}")
    print("-" * 25)
```
