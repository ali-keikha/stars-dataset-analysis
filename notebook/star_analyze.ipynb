Research Question:
Is there a relationship between a star's color and its temperature based on the visible light spectrum?

Some stars do not follow this expected relationship. We will investigate whether these groups contain fewer samples than usual and determine if the small sample size could explain their unusual behavio.

Checking for Duplicate Values

```python
colors = df["Color"].value_counts()
```

Standardizing Color Names

```python
df["Color"] = df["Color"].replace(
    {
    "Blue-white": "Blue White",
    "Blue white": "Blue White",
    "Blue-White": "Blue White",

    "white": "White",

    "yellowish": "Yellowish",

    "yellow-white": "Yellow White",

    "White-Yellow": "White Yellow",

    "Pale yellow orange": "Pale Yellow Orange",

    "Orange-Red": "Orange Red"
}
)
```

```python


temperature = []

for color in colors.index:
    temperature_mean = df[df["Color"] == color]["Temperature"].mean()
    temperature.append({
        "Color": color,
        "Temperature": temperature_mean
    })

temperature = pd.DataFrame(temperature)
    
```

Converting Colors to Wavelength Order

```python
color_order = {
    "Red": 1,
    "Orange Red": 2,
    "Orange": 3,
    "Pale Yellow Orange": 4,
    "Yellowish": 5,
    "Yellow White": 6,
    "White Yellow": 7,
    "Yellowish White": 8,
    "White": 9,
    "Whitish": 10,
    "Blue White": 11,
    "Blue": 12
}
temperature["Energy Rank"] = temperature["Color"].map(color_order)
```

```python
temperature.sort_values("Temperature")
```

Conclusion
In general, stars with higher temperatures tend to have colors closer to blue, while cooler stars are closer to red. Some exceptions were found, but these groups contain fewer samples. Therefore, the overall trend is still consistent with the expected relationship between color and temperature.
