# Star Dataset Analysis

```python
import pandas as pd
import numpy as np
```

## Research Question

Is there a relationship between a star's radius and its temperature?

## Step 1
##### Load Dataset

```python
df = pd.read_csv("../data/Stars.csv.xls")
```

## Step 2

#### Calculate the correlation between radius and temperature

```python
df[["Temperature", "R"]].head()
df[["Temperature", "R"]].describe()
df[["Temperature", "R"]].info()
df[["Temperature", "R"]].isnull().sum()
```

## Step 3

#### Calculate the correlation between radius and temperature

```python
df[["Temperature", "R"]].corr()
```

## Conclusion

The correlation coefficient between radius and temperature is 0.064, indicating a very weak positive relationship. Based on this dataset, there is no strong linear correlation between a star's radius and its temperature.
