## Basic Statistics

# Minimum temperature & Maximum temperature & Average temperature

```python
print(f"Minimum temperature: {df["Temperature"].min()} K")

print(f"Maximum temperature: {df["Temperature"].max()} K")

mean = df["Temperature"].mean()
print(f"erage temperature: {np.round(mean, 3)}")
```

# Minimum luminosity & Maximum luminosity & Average luminosity

```python
min_luminositydf = df["L"].min()
print(f"Minimum luminosity: {min_luminositydf}")

max_luminositydf = df["L"].max()
print(f"maximum luminosity: {max_luminositydf}")

print(f"Average luminosity: {np.round(df["L"].mean(), 3)}")
```

# Minimum radius & Maximum radius & Average radius

```python
print(f"Minimum radius: {df["R"].min()}")

print(f"Maximum radius: {df["R"].max()}")

print(f"Average radius: {np.round(df["R"].mean(), 3)}")
```

## Category Distributions

# Star Type counts

```python
df["Type"].value_counts()
```

# Star Color counts

```python
df["Color"].value_counts()
```

Observation:
Some star colors are written in different formats. These inconsistencies may affect the analysis, so they should be standardized.

```python
df["Color"] = df["Color"].replace(
    {
    "Blue-white": "Blue White",
    "Blue white": "Blue White",
    "Blue-White": "Blue White",

    "white": "White",

    "yellowish": "Yellowish",

    "yellow-white": "Yellow White",

    "White-Yellow": "White Yellow",

    "Pale yellow orange": "Pale Yellow Orange",

    "Orange-Red": "Orange Red"
}
)
```

```python
df["Color"].value_counts()
```
